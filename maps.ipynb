```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import requests
import plotly.graph_objects as go

# 1. Create the input text box and button
place_input = widgets.Text(
    value='Cubbon Park, Bangalore',
    placeholder='Enter place name (e.g. Bangalore Palace)',
    description='Place:',
    disabled=False,
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='400px')
)

search_button = widgets.Button(
    description='Search & Map',
    button_style='success',
    icon='search'
)

# Output container for the map and text results
output = widgets.Output()

def on_search_button_clicked(b):
    with output:
        clear_output(wait=True)
        query = place_input.value.strip()
        
        if not query:
            print("⚠️ Please enter a valid place name.")
            return
            
        print(f"🔍 Searching live maps for '{query}'...")
        
        # Use OpenStreetMap Nominatim API to convert place name to coordinates (Free, no API key needed)
        url = f"https://nominatim.openstreetmap.org/search?q={query}&format=json&limit=1"
        headers = {'User-Agent': 'JupyterNavigatorBot/1.0'}
        
        try:
            response = requests.get(url, headers=headers)
            data = response.json()
            
            if data:
                lat = float(data[0]['lat'])
                lon = float(data[0]['lon'])
                display_name = data[0]['display_name']
                
                print(f"✅ Found! {display_name}")
                print(f"Coordinates -> Latitude: {lat}, Longitude: {lon}\n")
                
                # Build the Plotly Mapbox centered on the searched location
                fig = go.Figure(go.Scattermapbox(
                    lat=[lat],
                    lon=[lon],
                    mode='markers+text',
                    marker=dict(size=22, color='crimson', symbol='star'),
                    text=[query],
                    textposition="top right"
                ))
                
                fig.update_layout(
                    mapbox=dict(
                        style="open-street-map",
                        center=dict(lat=lat, lon=lon),
                        zoom=15
                    ),
                    margin=dict(l=0, r=0, t=40, b=0),
                    title=f"📍 Live Map Result: {query}"
                )
                
                fig.show()
            else:
                print(f"❌ Could not find '{query}'. Try adding ', Bangalore' for better precision.")
                
        except Exception as e:
            print(f"⚠️ Network or API error: {e}")

# Link button click event to our search function
search_button.on_click(on_search_button_clicked)

# 3. Display the interactive form interface
print("🧭 Navigator Control Panel Loaded. Enter a destination below:")
display(widgets.VBox([place_input, search_button, output]))
```
